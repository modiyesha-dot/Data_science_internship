```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("career_readiness_cleaned.csv")
data.head()
```

```python
print("CAREER READINESS DASHBOARD")
print("-" * 30)

print("Total Students:", data["Student_Name"].count())
print("Average CGPA:", round(data["CGPA"].mean(), 2))
print("Maximum CGPA:", data["CGPA"].max())
print("Minimum CGPA:", data["CGPA"].min())
```

```python
avg_scores = data[
    ["Python","SQL","PowerBI","Communication","Aptitude"]
].mean()

plt.figure(figsize=(15,5))

# Bar Chart
plt.subplot(1,3,1)
plt.bar(avg_scores.index, avg_scores.values)
plt.title("Average Skill Scores")
plt.xticks(rotation=45)

# Line Chart
plt.subplot(1,3,2)
plt.plot(data.index, data["CGPA"], marker='o')
plt.title("CGPA Trend")
plt.xlabel("Students")
plt.ylabel("CGPA")

# Pie Chart
plt.subplot(1,3,3)
plt.pie(
    avg_scores.values,
    labels=avg_scores.index,
    autopct="%1.1f%%"
)
plt.title("Skill Distribution")

plt.tight_layout()
plt.show()
```

```python
print("KEY INSIGHTS")
print("-" * 30)

print("Highest Average Skill:",
      avg_scores.idxmax())

print("Lowest Average Skill:",
      avg_scores.idxmin())

print("Average CGPA:",
      round(data["CGPA"].mean(),2))

print("Total Student Records:",
      len(data))
```
